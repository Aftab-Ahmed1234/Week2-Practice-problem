**Count**

```python
text=input("Enter a string:  ")
count=0
for ch in text.lower():
    if ch in "aeiou":
       count +=1
print("output",count)
```

**Reverse Text**

```python
text=input("Enter String:  ")
reverse_text=text[::-1]
print("output", reverse_text)
```

**Palindrome**

```python
text=input("Enter String: ").lower()
if text==text[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")
```
